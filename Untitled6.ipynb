```python
"""
TUGAS MACHINE LEARNING
Implementasi Support Vector Machine (SVM) dalam Python

Nama : Wahyudi Amanda Sihombing
NPM  : 202425006

Materi mengacu pada video:
https://youtu.be/efR1C6CvhmE

Program ini memperlihatkan ide utama SVM:
1. Membuat dua kelompok data klasifikasi.
2. Membagi data menjadi data latih dan data uji.
3. Melakukan standardisasi fitur.
4. Melatih model SVM dengan kernel linear.
5. Menampilkan akurasi dan confusion matrix.
6. Menggambar hyperplane, margin, dan support vectors.

Instalasi pustaka:
pip install numpy matplotlib scikit-learn
"""

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


def main():
    # 1. Membuat contoh dataset dua kelas agar mudah divisualisasikan.
    X, y = make_blobs(
        n_samples=120,
        centers=2,
        n_features=2,
        cluster_std=1.35,
        random_state=42
    )

    # 2. Membagi data: 75% untuk pelatihan dan 25% untuk pengujian.
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.25, random_state=42, stratify=y
    )

    # 3. Standardisasi penting karena SVM sensitif terhadap skala fitur.
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # 4. Membuat dan melatih model Support Vector Machine.
    model = SVC(kernel="linear", C=1.0)
    model.fit(X_train_scaled, y_train)

    # 5. Melakukan prediksi pada data uji.
    y_pred = model.predict(X_test_scaled)

    print("=" * 55)
    print("HASIL KLASIFIKASI SUPPORT VECTOR MACHINE (SVM)")
    print("=" * 55)
    print(f"Jumlah data keseluruhan : {len(X)}")
    print(f"Jumlah data latih       : {len(X_train)}")
    print(f"Jumlah data uji         : {len(X_test)}")
    print(f"Jumlah support vectors  : {len(model.support_vectors_)}")
    print(f"Akurasi model           : {accuracy_score(y_test, y_pred) * 100:.2f}%")
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

    # 6. Visualisasi garis pemisah (hyperplane), margin, dan support vectors.
    x_min, x_max = X_train_scaled[:, 0].min() - 1, X_train_scaled[:, 0].max() + 1
    y_min, y_max = X_train_scaled[:, 1].min() - 1, X_train_scaled[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 400),
        np.linspace(y_min, y_max, 400)
    )

    grid = np.c_[xx.ravel(), yy.ravel()]
    decision_values = model.decision_function(grid).reshape(xx.shape)

    plt.figure(figsize=(9, 6))
    plt.contourf(
        xx, yy, decision_values,
        levels=[-100, 0, 100],
        alpha=0.15
    )
    plt.contour(
        xx, yy, decision_values,
        levels=[-1, 0, 1],
        linestyles=["--", "-", "--"],
        linewidths=[1.5, 2.5, 1.5]
    )

    plt.scatter(
        X_train_scaled[:, 0], X_train_scaled[:, 1],
        c=y_train, cmap="coolwarm", edgecolors="black",
        s=55, label="Data latih"
    )
    plt.scatter(
        model.support_vectors_[:, 0],
        model.support_vectors_[:, 1],
        s=180, facecolors="none", edgecolors="black",
        linewidths=1.8, label="Support vectors"
    )

    plt.title("Klasifikasi Support Vector Machine (SVM) - Linear")
    plt.xlabel("Fitur 1 (setelah standardisasi)")
    plt.ylabel("Fitur 2 (setelah standardisasi)")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.savefig("hasil_visualisasi_svm.png", dpi=300)
    plt.show()

    # 7. Contoh prediksi satu data baru.
    data_baru = np.array([[0.0, 0.0]])
    data_baru_scaled = scaler.transform(data_baru)
    hasil = model.predict(data_baru_scaled)[0]
    print(f"\nPrediksi untuk data baru [0.0, 0.0]: Kelas {hasil}")
    print("Grafik disimpan sebagai: hasil_visualisasi_svm.png")


if __name__ == "__main__":
    main()
```
